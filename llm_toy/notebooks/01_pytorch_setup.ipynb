# PyTorch 设置与 GPU 验证

本 Notebook 验证 PyTorch 是否正确安装，并确保其能够访问您的 RTX 2070 GPU。

```python
import torch
import torch.nn as nn
import numpy as np
import sys

print(f"Python 版本: {sys.version}")
print(f"PyTorch 版本: {torch.__version__}")
print(f"CUDA 可用: {torch.cuda.is_available()}")
print(f"CUDA 版本: {torch.version.cuda}")
```

```python
# 检查 GPU 详情
if torch.cuda.is_available():
    print(f"GPU 数量: {torch.cuda.device_count()}")
    for i in range(torch.cuda.device_count()):
        print(f"GPU {i}: {torch.cuda.get_device_name(i)}")
        print(f"已分配内存: {torch.cuda.memory_allocated(i) / 1024**3:.2f} GB")
        print(f"已缓存内存: {torch.cuda.memory_reserved(i) / 1024**3:.2f} GB")
        print(f"总内存: {torch.cuda.get_device_properties(i).total_memory / 1024**3:.2f} GB")
        print("-" * 50)
```

```python
# 测试 GPU 计算
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"使用设备: {device}")

# 在 GPU 上创建张量
x = torch.randn(1000, 1000, device=device)
y = torch.randn(1000, 1000, device=device)

# 执行矩阵乘法
z = torch.matmul(x, y)
print(f"矩阵乘法结果形状: {z.shape}")
print(f"计算设备: {z.device}")
```

```python
# 测试 GPU 内存使用
if torch.cuda.is_available():
    print("创建大张量前:")
    print(f"已分配内存: {torch.cuda.memory_allocated() / 1024**3:.2f} GB")
    print(f"已缓存内存: {torch.cuda.memory_reserved() / 1024**3:.2f} GB")
    
    # 创建一个大张量
    large_tensor = torch.randn(5000, 5000, device=device)
    
    print("\n创建大张量后:")
    print(f"已分配内存: {torch.cuda.memory_allocated() / 1024**3:.2f} GB")
    print(f"已缓存内存: {torch.cuda.memory_reserved() / 1024**3:.2f} GB")
    
    # 删除张量并清空缓存
    del large_tensor
    torch.cuda.empty_cache()
    
    print("\n清理后:")
    print(f"已分配内存: {torch.cuda.memory_allocated() / 1024**3:.2f} GB")
    print(f"已缓存内存: {torch.cuda.memory_reserved() / 1024**3:.2f} GB")
```
